```python
# 训练词向量
# 1.从xml文本里面提取内容
def strip_tag(infile,outfile):
    import re

    # 1. 读取文件
    with open(infile, 'r', encoding='utf-8') as f:
        content = f.read()

    # 2. 用正则清除标签
    clean_text = re.sub(r'<[^>]+>', '', content)

    # 3. 写入新文件
    with open(outfile, 'w', encoding='utf-8') as f:
        f.write(clean_text)
```

```python
strip_tag('enwik9/enwik9',"enwik9_no_tag")
```
